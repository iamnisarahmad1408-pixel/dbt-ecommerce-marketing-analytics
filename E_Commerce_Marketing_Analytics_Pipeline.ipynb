```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

np.random.seed(42)
dates = [datetime(2026, 7, 1) + timedelta(days=i) for i in range(30)]

# 1. Generate Raw Ad Spend Data
ad_platforms = ['meta', 'google']
campaigns = ['CAMP_SUMMER_01', 'CAMP_RETARGETING_02', 'CAMP_BRAND_03']

ad_spend_rows = []
for d in dates:
    for platform in ad_platforms:
        for camp in campaigns:
            ad_spend_rows.append({
                'spend_date': d.strftime('%Y-%m-%d'),
                'platform': platform,
                'campaign_id': f"{platform.upper()}_{camp}",
                'impressions': np.random.randint(1000, 10000),
                'clicks': np.random.randint(50, 500),
                'spend_usd': round(np.random.uniform(50.0, 500.0), 2)
            })

df_ad_spend = pd.DataFrame(ad_spend_rows)
df_ad_spend.to_csv('raw_ad_spend.csv', index=False)

# 2. Generate Raw Shopify Orders Data
order_rows = []
order_id = 1001

for d in dates:
    daily_orders = np.random.randint(20, 50)
    for _ in range(daily_orders):
        channel = np.random.choice(['meta', 'google', 'organic', 'email'], p=[0.4, 0.3, 0.2, 0.1])
        subtotal = round(np.random.uniform(25.0, 250.0), 2)
        discount = round(subtotal * np.random.choice([0.0, 0.1, 0.2], p=[0.7, 0.2, 0.1]), 2)

        order_rows.append({
            'order_id': f"ORD_{order_id}",
            'customer_id': f"CUST_{np.random.randint(100, 500)}",
            'created_at': (d + timedelta(hours=np.random.randint(0, 23), minutes=np.random.randint(0, 59))).strftime('%Y-%m-%d %H:%M:%S'),
            'subtotal_usd': subtotal,
            'discount_usd': discount,
            'net_revenue_usd': subtotal - discount,
            'utm_source': channel
        })
        order_id += 1

df_orders = pd.DataFrame(order_rows)
df_orders.to_csv('raw_shopify_orders.csv', index=False)

print("Data generation complete: 'raw_ad_spend.csv' and 'raw_shopify_orders.csv' created.")
```
